```python
import os

# Set these before importing laya: Colab has TensorFlow installed by default,
# which can deadlock model loading if not bypassed. Disable tokenizer parallel forks.
os.environ.setdefault("USE_TF", "0")
os.environ.setdefault("USE_TORCH", "1")
os.environ.setdefault("TOKENIZERS_PARALLELISM", "false")

import laya

# Pulls the open-weights model directly from Hugging Face Hub into the Colab environment
agent = laya.load("convaiinnovations/laya")

result = agent.predict(
    {"body": "Hi, we were billed twice for March. Please refund the duplicate today."},
    {"refund_requested": {
        "type": "noul",
        "instructions": "Does the user explicitly request a refund?",
    }},
)

answer = result["answers"]["refund_requested"]
print("P(refund requested) = %.4f  (confidence %.4f)" % (answer["noul"], answer["confidence"]))
```
